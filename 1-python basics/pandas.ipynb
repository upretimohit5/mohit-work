```python
## series- it is a one-dimentional array like object that can hold any data type. it is similar to column in a table

import pandas as pd
data=[1,2,3,4]
series=pd.Series(data)
print("series \n", series)
```

```python
## creat a series from dictionary
data={'a':1,'b':2, 'c':3}
seires_dict=pd.Series(data)
print(seires_dict)
print(type(seires_dict))
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index)
```

```python
## Data frame
## creat a Dataframe from a dictionary of list

data={
  'Name':['Mohit', 'Rohit', 'Mayank'],
  'Age':[26,20,26],
  'city':['dub', 'gur', 'rai']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## creat a data frame form a list of dictionary
data=[
  {'Name': 'Mohit', 'Age': 26, 'City':'dublin'},
  {'Name': 'Rohit', 'Age': 20, 'City':'dublin'},
  {'Name': 'Mayank', 'Age': 26, 'City':'dublin'},
  {'Name': 'Neha', 'Age': 23, 'City':'dublin'}
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```
