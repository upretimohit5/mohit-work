```python
##the year is leap year or not
year = 2020

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
        print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

```python
## claculate the first n natural number using while and for loop
## while loop
n=10
sum=0
count=1
while count<=n:
  sum=sum+count
  count=count+1
  
print("the sum is of n natural number:", sum)
```
