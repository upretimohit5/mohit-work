```python
## creat a set
my_set = {1,2,3,4,5}
print(type(my_set))
```

```python
## my empty set
empty_set = set()
print(type(empty_set))
```

```python
## duplicate values will be ignored in the set
dup_set = {1,2,3,4,5,1,2,3}
print(dup_set) ## it will print {1,2,3,4,5} only
```

```python
## add elements to a set
dup_set.add(6)
print(dup_set)
```

```python
## rmove elements from a set
dup_set.remove(3) ## if the element is not present it will throw an error
print(dup_set)
```

```python
## discard basiically same as remove but if the element is not present it will not throw an error
dup_set.discard(10) ## it will not throw an error
print(dup_set)
```

```python
## pop method removes and returns an arbitrary set element.
removed_element = dup_set.pop()
print(removed_element) ## it will remove first element
print(dup_set)
```

```python
## delete or clear all the elements from the set
dup_set.clear()
print(dup_set)
```

```python
## to check element is present in the set or not
check_set = {1,2,3,4,5}
print(3 in check_set) ## it will return True
print(10 in check_set) ## it will return False
```

```python
## mathematical set operations
A = {1,2,3,4,5}
B = {4,5,6,7,8}

### union of two sets
union_set = A.union(B)
print(union_set)

### intersection of two sets
intersection_set = A.intersection(B)
print(intersection_set)

## intersection update to give the intersection value to the first set
A.intersection_update(B) ## A will now have the intersection value
print(A)
```

```python
A = {1,2,3,4,5}
B = {4,5,6,7,8}
### difference of two sets
diffrence_set = A.difference(B)
print(diffrence_set)
```

```python
## symmetric difference of two sets
symmetric_diff_set = A.symmetric_difference(B) ## it will give the elements which are not common in both sets
print(symmetric_diff_set)
```

```python
## subset and superset
A = {1,2,3,4,5}
B = {1,2,3}
print(B.issubset(A)) ## it will return True becase B is subset of A
print(B.issuperset(A)) ## it will return False because B is not superset of A
```

```python
text = "In this tutorial we did discuss about the text in notebook"
words = text.split()
## convert list to set to get the unique words
unique_words = set(words)
print(unique_words)
print(len(unique_words))
```
