```python
### addition in lambda
addition = lambda a,b:a+b
type(addition)  ## this was just for the type of so gives function because lambda is a function
print(addition(10,12))
```

```python
even= lambda num:num%2==0
num = 13
if even(num):
  print(f"{num} is even number")
else:
  print(f"{num} is odd number")

```

```python
## example of lambda in main thing
number = [1,2,3,4,5,6]
eve = lambda n:n%2==0
even_number = list(filter(eve,number))  ##Because filter needs to know which collection (list) it should go through. If you only wrote filter(eve), Python would say: “Okay... but what do I filter? 🤔” So the number is the data source, and eve is the rule for what to keep
print(even_number)  
```
