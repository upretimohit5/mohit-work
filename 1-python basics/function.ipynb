```python
## why function?
# 1. code reuseability
# 2. modularity - breaking a complex problem into smaller sub-problems
# 3. readability - makes code easier to read and understand
# 4. maintainability - easier to update and maintain code

def even_odd(num):
  """this function checks if a number is even or odd"""
  if num%2==0:
    return"even"
  else:
    return"odd"
print(even_odd(4))
print(even_odd(7))
print(even_odd(10))
print(even_odd(15))
```

```python
## variable length arguments
## position and keyword arguments

def print_numbers(*args): ## as a good coder we always use args
  """this function prints all the numbers passed to it"""
  for num in args:
    print(num)
```

```python
print_numbers(5,5,3,5,7,8,9,1,2,3,4,5,"Mohit")
```

```python
### keyword arguments

def print_details(**kwargs):
  """this function prints all the details passed to it"""
  for key,value in kwargs.items():
   print(f"{key} : {value}")
  
```

```python
print_details(name="Mohit Upreti", age="26", city="delhi", country="India")
```

```python
## positional and keyword arguments can be used together
def display_info(*args, **kwargs):
  """this function displays all the info passed to it"""
  for number in args:
    print(f"positional argumanet: {number}")
  
  for key,value in kwargs.items():
    print(f"{key} : {value}")

```

```python
display_info(1,2,3,4,5,"Mohit",name="Mohit Upreti", age="26", city="delhi", country="India")
```
