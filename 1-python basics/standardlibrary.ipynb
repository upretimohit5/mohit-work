```python
## Data Serialization
import json
data= {'name': 'Mohit Upreti', 'age':27}
print(type(data))

json_str = json.dumps(data) ## it convert data in string form
print(json_str)
print(type(json_str))

parsed_data=json.loads(json_str) ## it convert data in dictionary form
print(parsed_data)
print(type(parsed_data))

```

```python
## csv
import csv
from operator import ne  ## to make file comma seprated values bsically in tabular data in plain text
with open('example.csv',mode='w',newline='') as file:  ## mode w is use for write the file
  writer = csv.writer(file)
  writer.writerow(['name', 'age'])
  writer.writerow(['Mohit Upreti', 27])

with open('example.csv',mode='r') as file: ## mode r is use for read the file
  reader= csv.reader(file)
  for row in reader:
    print(row)
```
