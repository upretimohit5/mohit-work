### Data Encoding

#### this is just for to change the value from categories to numerical value to train the model because the machine need the value likke to understand the value in numerices

##### there are three types of data encoding

##### 1.nominal encoding/one hot encoding(OHE)   2.label and ordinal encoding     3.target guided ordinal encoding

```python
### one hot encoding(OHE)
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder
```

```python
## creat a sample data frame
df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'green', 'red', 'blue', ]
})
```

```python
df.head()
```

```python
### create an OneHotEncoder instance
encoder=OneHotEncoder()
```

```python
### perform fit and then transform
encoded=encoder.fit_transform(df[['color']]).toarray()   ## sparse Matrix is used to eleminate the zero from the the data to make the memory lightweight
```

```python
import pandas as pd
encoder_df = pd.DataFrame(encoded,columns=encoder.get_feature_names_out(['color'])) 
```

```python
encoded
```

```python
encoder_df
```

```python
## for any ner data
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```

### ordinal encoding

```python
### it is used for to put data in order
### ordinal encoding
from sklearn.preprocessing import OrdinalEncoder
```

```python
## creat a sample data frame
df = pd.DataFrame({
    'size': ['small', 'medium', 'large', 'medium', 'small', 'large']
}) 
```

```python
df
```

```python
## creat an instance of OrdinalEncoder adn then fit_transformation
encoder= OrdinalEncoder(categories=[['small', 'medium', 'large']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
encoder.transform([['medium']])
```
