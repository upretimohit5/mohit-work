### Handling imbalanced dataset

###### 1.Up Sampling
###### 2.Down Sampling

```python
import numpy as np
import pandas as pd
## set the random seed for reproducibility
np.random.seed(123)

## create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0, n_class_1
```

```python
## create my dataframe with imbalanced dataset
class_0 = pd.DataFrame({
    'feature1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target':[0]*n_class_0
})  

class_1 = pd.DataFrame({
    'feature1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target':[1]*n_class_1
})
```

```python
df = pd.concat([class_0, class_1]).reset_index(drop=True)  ##Reseta_index work as fix the 
```

```python
df.head()
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
## upsampling
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample
df_minority_upsampled = resample(df_minority,replace=True, ##Sample with replacement
         n_samples=len(df_majority), ##to match majority class
         random_state=42
        ) ##reproducible results
```

```python
## downsapmling
## set the random seed for reproducibility
np.random.seed(123)

## create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0

## create my dataframe with imbalanced dataset
class_0 = pd.DataFrame({
    'feature1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target':[0]*n_class_0
})  

class_1 = pd.DataFrame({
    'feature1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target':[1]*n_class_1
})

df = pd.concat([class_0, class_1]).reset_index(drop=True)  ##Reset_index work as fix the numbering after concat

## check the class distrubation
print(df['target'].value_counts())

```

```python
## downsampling
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample
df_majority_downsampled = resample(df_majority,replace=False,
         n_samples=len(df_minority), ##to match majority class
         random_state=42
        ) ##reproducible results
```

```python
df_majority_downsampled.shape
```

```python
df_downsampled=pd.concat([df_majority_downsampled, df_minority])
```
