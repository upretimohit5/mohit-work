### 5 number summary and box plot

```python
## Minimum, Maximum, Median, Q1(first quartile), Q3(third quartile)
```

```python
import numpy as np
```

```python
lst_marks=[45,32,56,75,89,90,54,32,89,23,67,88,92,100,34,23,45,67,78,88,99,43,56,78,90]
minimum,Q1,median,Q3,max=np.quantile(lst_marks,[0,0.25,0.50,0.75,1.0])
```

```python
minimum,Q1,median,Q3,max
```

```python
## find the outliers using IQR method
IQR=Q3-Q1
print("IQR is:",IQR)
```

```python
lower_fence=Q1-1.5*IQR
upper_fence=Q3+1.5*IQR
```

```python
lower_fence
```

```python
upper_fence
```

```python
import seaborn as sns
```

```python
sns.boxplot(lst_marks)
```
