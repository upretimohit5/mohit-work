```python
x= [1,2,3,4,5,6,7]
print(list(map(lambda x:x**2,x)))
```

```python
## map multiple iterable
number1= [1,2,3]
number2= [4,5,6]
print(list((map(lambda x,y:x+y,number1,number2))))
```

```python
## map() is use to convert string to list
str_numbers = ["1", "2", "3", "4", "5"]
for str_number in str_numbers:
  print(type(str_number))
int_numbers=list(map(int, str_numbers)) ## int here is inbuilt functio in map()
print(int_numbers)
for int_number in int_numbers:
  print(type(int_number))
```

```python
words = ["apple", "orange", "cherry"]
upper_word = list(map(str.upper, words)) ## more example of inbuilt function in here
print(upper_word)
```

```python
def get_name(person):
  return person['name']

people = [
  {'name':'Mohit Upreti', 'age' :27},
  {'name':'Rohit Upreti', 'age' :20}
]
list(map(get_name, people))
```
