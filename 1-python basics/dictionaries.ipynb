```python
## dictiocnary with some values key:value pairs
student = {'name': 'Mohit', 'age':24, 'course':'Data Science'}
print(student)
print(type(student))
```

```python
## single key always used only once in the dictionary
student = {'name': 'Mohit', 'age':24, 'course':'Data Science', 'name':'Rohit'}
print(student) ## it will consider the last key value pair if the key is repeated
```

```python
## accessing the value of a key in the dictionary
student = {'name': 'Mohit', 'age':24, 'course':'Data Science'}
print(student['course']) ## it will give the value of the key course
```

```python
## shallow copy of dictionary
student = {'name': 'Mohit', 'age':24, 'course':'Data Science'}
student_copy = student
print(student_copy)
print(student)
```

```python
student['name'] = 'Rohit' ## changing the name in the original dictionary
print(student_copy)
print(student)
```

```python
student_copy1 = student.copy() ## shallow copy
print(student_copy1)
print(student)
```

```python
student['name'] = 'Upreti'
print(student_copy1)
print(student) ## it is possible because of shallow copy
```

```python
### iterating through a dictionary
## you can use loops to iterate

## iteratinf over keys
for key in student.keys():
  print(key)
```

```python
## iterate over values
for value in student.values():
  print(value)
```

```python
## iterate over key value pairs
for key,value in student.items():
  print (f"{key} : {value}")
```

```python
## nested dictionaries
students = {
  "student1" : {'name': 'Mohit', 'age':24},
  "student2" : {'name': 'Rohit', 'age':21}
}
print(students)
```

```python
## access nested dictionary vlaue elements
print(students["student1"]["name"])
print(students["student1"]["age"])
```

```python
## iterating over nested dictionaries
for student_key, student_info in students.items(): ## here is student_key = student1, student2 and student_info is the nested dictionary
  print(f"{student_key} : {student_info}")
  for key, value in student_info.items(): ##
    print(f"{key} : {vlaue}")
```

```python
## dictionary comprehension
## long method
## for x in range(5):
## square = x**2
## print(f"the square of {x} is {square}")

## main method
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## condition in dictionary comprehension
even_squares={x:x**2 for x in range (10) if x%2==0}
print(even_squares)
```

```python
## practicle example
##creat a dictionary to count the number of the frequency in the given list
numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}
print(type(frequency))

for number in numbers:
  if number in frequency:
      frequency[number]+= 1
  else:
      frequency[number] = 1
print(frequency)
```

```python
## how many times each word appears in the list
words = ["apple", "banana", "apple", "orange", "banana", "apple"]
word_count = {}

for word in words:
  if word in word_count:
    word_count[word] += 1
  else:
    word_count[word] = 1
print(word_count)
```

```python
##find which numbers appear only once
values= [5, 8, 2, 5, 7, 8, 5, 2]
frequency = {}

for value in values:
  if value in frequency:
    frequency[value] +=1
  else:
    frequency[value] =1
print(frequency)

unique_number = []
for key,lock in frequency.items():
  if lock == 1:
    unique_number.append(key)
print(unique_number)

```
