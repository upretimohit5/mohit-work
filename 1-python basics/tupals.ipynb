```python
## tuple methods- basically there are we are goona use count and index method
number = (1, 2, 3, 4, 5, 1, 2, 1)
print(number.count(1))  ## it will count how many time 1 is there in the tuple
print(number.index(3)) ## it will give the index of 3
```

```python
## packing and unpacking of tuples
## paking
packed_tuple = 1, 2, "hello", 3.14
print(packed_tuple) ## by default it will be in tuple form
```

```python
# unpacking tuple
a,b,c,d = packed_tuple
print(a)
print(b)
print(c)
print(d) ##here we are unpacking the tuple into individual variables
```

```python
## unpacking with asterisk(*)
number = (1, 2, 3, 4, 5, 6, 7, 8, 9)
first, *middle, last = number
print(first)
print(middle) ## this is because of asterisk all the middle values will be in the form of list
print(last) 
```

```python
## nested tuples
nested_tuple = (1,2,(3,4),(5,6))
print(nested_tuple)
print(nested_tuple[2]) ## it will print (3,4)
print(nested_tuple[2][1]) ## it will print 4
```
