#### Example of to manage a to do list

##### *creat a to do list application using list in python

```python
to_do_list = ["Buy groceries", "clean the house", "pay bills"]

## adding to task
to_do_list.append("secdule meeting")
to_do_list.append("call mom")

## removing a task
to_do_list.remove("clean the house")

## checking if a task is in the list
if "pay bills" in to_do_list:
  print("you have to pay bills")

  print("to do list remaining tasks:")
  for task in to_do_list:
    print(f"- {task}")
```

## Example-orgainizing student grades

##### creat a list to store and calculate avarage grades for students

```python
#organizing student grades
grades = [85, 92, 78, 90, 88]

## adding a new grade
grades.append(95)

## claculating the avarage grades
average_grades = sum(grades) / len(grades)
print(f"the avarage grades is: {average_grades}")

## find the highest and lowest grades
highest_grade = max(grades)
lowest_grade = min(grades)
print(f"the highest grade is: {highest_grade}")
print(f"the lowest grade is: {lowest_grade}")
```

## Managing the inventory.

##### used a list to manage inventory item in a store.

```python
## managing the inventory.
inventory = ["apple", "banana", "orange"]

## adding new items to the inventory
inventory.append("grapes")

## removing an item that is out of stock
inventory.remove("banana")

## checking if an item is in stock
item = "mango"
if item in inventory:
  print(f"{item} is in stock")
else:
    print(f"{item} is out of stock")

print("current inventory:")
for item in inventory:
  print(f"- {item}")
```

## collecting user feedback.

##### use a list to collect and analyze user feedback

```python
## collection fo user feedback.
feedback = ["great service", "friendly staff", "could be better", "excellent quality"]

## adding new feedback
feedback.append("average experience")

## analyzing spefecific feedback
positive_feedback_count = sum(1 for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower())
print(f"postive_feedback_count: {positive_feedback_count}")

## print all feedback
print("user feedabck:")
for comment in feedback:
  print(f"- {comment}")
```
